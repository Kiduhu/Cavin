Below is the code created by your colleague to calculate the feature described above. Use this code to re-create this feature and then think about ways to build on this feature to create features with a higher predictive power.

```python
price_df = pd.read_csv('/content/price_data (1).csv')
price_df["price_date"] = pd.to_datetime(price_df["price_date"], format='%Y-%m-%d')
price_df.head()
```

```python
# Group off-peak prices by companies and month
monthly_price_by_id = price_df.groupby(['id', 'price_date']).agg({'price_off_peak_var': 'mean', 'price_off_peak_fix': 'mean'}).reset_index()

# Get january and december prices
jan_prices = monthly_price_by_id.groupby('id').first().reset_index()
dec_prices = monthly_price_by_id.groupby('id').last().reset_index()

# Calculate the difference
diff = pd.merge(dec_prices.rename(columns={'price_off_peak_var': 'dec_1', 'price_off_peak_fix': 'dec_2'}), jan_prices.drop(columns='price_date'), on='id')
diff['offpeak_diff_dec_january_energy'] = diff['dec_1'] - diff['price_off_peak_var']
diff['offpeak_diff_dec_january_power'] = diff['dec_2'] - diff['price_off_peak_fix']
diff = diff[['id', 'offpeak_diff_dec_january_energy','offpeak_diff_dec_january_power']]
diff.head()
```

# 1. Tenure-Related Features

```python
# Calculate contract duration and days since last renewal/modification
df['contract_duration'] = (df['date_end'] - df['date_activ']).dt.days
df['days_since_renewal'] = (df['date_end'] - df['date_renewal']).dt.days
df['days_since_modif_prod'] = (df['date_end'] - df['date_modif_prod']).dt.days
```

# 2. Consumption Features

```python
# Recent consumption ratio (last month vs monthly average)
df['monthly_avg_cons'] = df['cons_12m'] / 12
df['recent_cons_ratio'] = np.where(df['monthly_avg_cons'] != 0,
                                   df['cons_last_month'] / df['monthly_avg_cons'],
                                   0)

# Gas consumption percentage
df['gas_percentage'] = np.where(df['cons_12m'] != 0,
                               df['cons_gas_12m'] / df['cons_12m'],
                               0)
```

```python
df.tail()
```

# 3. Advanced Price Features

```python
# Aggregate price statistics for each customer
price_agg = price_df.groupby('id').agg({
    'price_off_peak_var': ['mean', 'std', 'max', 'min'],
    'price_peak_var': ['mean', 'std'],
    'price_off_peak_fix': ['mean', 'std'],
}).reset_index()

# Flatten multi-index columns
price_agg.columns = ['id',
                    'offpeak_var_mean', 'offpeak_var_std', 'offpeak_var_max', 'offpeak_var_min',
                    'peak_var_mean', 'peak_var_std',
                    'offpeak_fix_mean', 'offpeak_fix_std']

# Merge with main DataFrame
df = pd.merge(df, price_agg, on='id', how='left')

# Price change over time (last - first)
price_sorted = price_df.sort_values(['id', 'price_date'])
first_prices = price_sorted.groupby('id').first().reset_index()
last_prices = price_sorted.groupby('id').last().reset_index()

price_change = pd.merge(
    first_prices[['id', 'price_off_peak_var', 'price_off_peak_fix']],
    last_prices[['id', 'price_off_peak_var', 'price_off_peak_fix']],
    on='id', suffixes=('_first', '_last')
)
price_change['offpeak_var_change'] = price_change['price_off_peak_var_last'] - price_change['price_off_peak_var_first']
price_change['offpeak_fix_change'] = price_change['price_off_peak_fix_last'] - price_change['price_off_peak_fix_first']

df = pd.merge(df, price_change[['id', 'offpeak_var_change', 'offpeak_fix_change']], on='id', how='left')
```
